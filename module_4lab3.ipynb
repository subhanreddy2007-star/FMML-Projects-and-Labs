 how changing the `batch_size` impacts how Mini-Batch Gradient Descent learns.

1.  A really small batch size (like 2):** This should act a lot like Stochastic Gradient Descent, with quite a bit of noise in its updates.
2.  A larger batch size (say, 80):** This should give us more stable updates than the tiny batches, but still be quicker than using the whole dataset.
3.  A batch size equal to the total number of samples (100):** When the batch size is the same as our entire dataset, this is essentially just standard Batch Gradient Descent.

```python
lr_mini_original = 0.1
n_iter_mini_original = 200
batch_size_original = 20

y_pred_mini_original = np.random.randn(2,1)
y_pred_mini_final_original, cost_history_mini_original = minibatch_gradient_descent(X, y, y_pred_mini_original, lr_mini_original, n_iter_mini_original, batch_size=batch_size_original)

print(f"Original Mini-Batch GD (batch_size={batch_size_original}):")
print(f'  y_pred[0]: {y_pred_mini_final_original[0][0]:0.3f}\ny_pred[1]: {y_pred_mini_final_original[1][0]:0.3f}')
print(f'  Final error: {cost_history_mini_original[-1]:0.3f}')
```

```python
lr_mini_small_batch = 0.1
n_iter_small_batch = 200
batch_size_small = 2

y_pred_small_batch = np.random.randn(2,1)
y_pred_mini_final_small_batch, cost_history_small_batch = minibatch_gradient_descent(X, y, y_pred_small_batch, lr_mini_small_batch, n_iter_small_batch, batch_size=batch_size_small)

print(f"Mini-Batch GD (batch_size={batch_size_small}):")
print(f'  y_pred[0]: {y_pred_mini_final_small_batch[0][0]:0.3f}\ny_pred[1]: {y_pred_mini_final_small_batch[1][0]:0.3f}')
print(f'  Final error: {cost_history_small_batch[-1]:0.3f}')
```

```python
lr_mini_large_batch = 0.1
n_iter_large_batch = 200
batch_size_large = 80

y_pred_large_batch = np.random.randn(2,1)
y_pred_mini_final_large_batch, cost_history_large_batch = minibatch_gradient_descent(X, y, y_pred_large_batch, lr_mini_large_batch, n_iter_large_batch, batch_size=batch_size_large)

print(f"Mini-Batch GD (batch_size={batch_size_large}):")
print(f'  y_pred[0]: {y_pred_mini_final_large_batch[0][0]:0.3f}\ny_pred[1]: {y_pred_mini_final_large_batch[1][0]:0.3f}')
print(f'  Final error: {cost_history_large_batch[-1]:0.3f}')
```

```python
lr_full_batch = 0.1
n_iter_full_batch = 200
batch_size_full = len(y)

y_pred_full_batch = np.random.randn(2,1)
y_pred_mini_final_full_batch, cost_history_full_batch = minibatch_gradient_descent(X, y, y_pred_full_batch, lr_full_batch, n_iter_full_batch, batch_size=batch_size_full)

print(f"Mini-Batch GD (batch_size={batch_size_full} - essentially Batch GD):")
print(f'  y_pred[0]: {y_pred_mini_final_full_batch[0][0]:0.3f}\ny_pred[1]: {y_pred_mini_final_full_batch[1][0]:0.3f}')
print(f'  Final error: {cost_history_full_batch[-1]:0.3f}')
```

the costs changed over time for these different batch sizes

```python
plt.figure(figsize=(12, 8))
plt.plot(range(n_iter_mini_original), cost_history_mini_original, 'm-', label=f'Original Mini-Batch GD (batch_size={batch_size_original})')
plt.plot(range(n_iter_small_batch), cost_history_small_batch, 'r-', label=f'Mini-Batch GD (batch_size={batch_size_small})')
plt.plot(range(n_iter_large_batch), cost_history_large_batch, 'g--', label=f'Mini-Batch GD (batch_size={batch_size_large})')
plt.plot(range(n_iter_full_batch), cost_history_full_batch, 'b:', label=f'Mini-Batch GD (batch_size={batch_size_full}, i.e., Batch GD)')

plt.xlabel('Number of iterations')
plt.ylabel('Cost')
plt.title('Comparison of Mini-Batch Gradient Descent with Different Batch Sizes')
plt.legend()
plt.grid(True)
plt.show()
```
