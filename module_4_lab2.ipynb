## Module 4: Perceptron and Gradient Descent
### Lab 2: Introduction to Gradient Descent

Ever wonder how machine learning models, especially those powerful deep learning networks, learn to make such accurate predictions? The secret sauce for most of them is a technique called **Gradient Descent**. It's absolutely foundational, and once you grasp it, a lot of other complex concepts start to click into place.

 Gradient Descent

Imagine you're lost on a foggy mountain, and your goal is to get down to the lowest point in the valley. You can't see far ahead, so what do you do? You'd probably feel the ground around you, figure out which way is the steepest *downhill*, and take a small step in that direction. You'd keep doing this, step by tiny step, until you can't go any lower. That's essentially Gradient Descent in a nutshell!

In the world of machine learning:
*   Our "mountain" is the **loss function**. Think of it as a measure of how badly our model is performing – how 'wrong' its predictions are.
*   The "lowest point" in the valley represents the ideal set of **model parameters** (or weights). This is where our model makes the fewest errors possible.

### The Algorithm in Action

Gradient Descent follows a pretty straightforward process:

1.  **Start Somewhere Random:** We begin by giving our model's parameters some initial, often random, values.
2.  **Find the Steepest Path Up:** Next, we calculate the **gradient** of our loss function. This gradient is like a compass that points directly towards the steepest *uphill* direction. If we were on the mountain, it would tell us which way to climb higher.
3.  **Take a Step Downhill:** Since we want to minimize our errors (go *down* the mountain), we move in the exact opposite direction of the gradient. We take a small, controlled step in that 'downhill' direction, adjusting our model's parameters.
4.  **Repeat and Refine:** We keep repeating steps 2 and 3 – calculating the gradient and taking a step – until our model's errors stop getting smaller. When the loss plateaus or stops decreasing significantly, we've likely found our minimum.



```python
import numpy as np
import random

np.random.seed(42)
random.seed(42)

X = np.linspace(-10, 10, 100)
true_a, true_b, true_c = 1, 0, 0
y_true = true_a * X**2 + true_b * X + true_c

noise = np.random.normal(0, 0.1, X.shape)
y = y_true + noise

def calculate_mse(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)

initial_a = np.random.rand() * 10
initial_b = np.random.rand() * 10
initial_c = np.random.rand() * 10

def calc_gradient_2nd_poly_for_GD(X, y_true, a, b, c):
    y_pred = a * X**2 + b * X + c
    N = len(X)

    da = (-2/N) * np.sum((y_true - y_pred) * X**2)
    db = (-2/N) * np.sum((y_true - y_pred) * X)
    dc = (-2/N) * np.sum(y_true - y_pred)

    return da, db, dc

def gradient_descent(X, y, initial_a, initial_b, initial_c, learning_rate, epochs):
    a, b, c = initial_a, initial_b, initial_c
    losses = []

    for epoch in range(epochs):
        da, db, dc = calc_gradient_2nd_poly_for_GD(X, y, a, b, c)

        a = a - learning_rate * da
        b = b - learning_rate * db
        c = c - learning_rate * dc

        y_pred = a * X**2 + b * X + c
        loss = calculate_mse(y, y_pred)
        losses.append(loss)

    return a, b, c, losses

learning_rate = 0.0003
epochs = 30000

print("Starting Gradient Descent training...")
final_a, final_b, final_c, losses = gradient_descent(X, y, initial_a, initial_b, initial_c, learning_rate, epochs)
print("Gradient Descent training complete.")
print(f"Initial coefficients: a={initial_a:.4f}, b={initial_b:.4f}, c={initial_c:.4f}")
print(f"Final learned coefficients: a={final_a:.4f}, b={final_b:.4f}, c={final_c:.4f}")
print(f"Final MSE loss: {losses[-1]:.4f}")
```

### Visualizing the Results of Gradient Descent

```python
import numpy as np
import matplotlib.pyplot as plt
import random

np.random.seed(42)
random.seed(42)

X = np.linspace(-10, 10, 100)
true_a, true_b, true_c = 1, 0, 0
y_true = true_a * X**2 + true_b * X + true_c

noise = np.random.normal(0, 0.1, X.shape)
y = y_true + noise

def calculate_mse(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)

initial_a = np.random.rand() * 10
initial_b = np.random.rand() * 10
initial_c = np.random.rand() * 10

def calc_gradient_2nd_poly_for_GD(X, y_true, a, b, c):
    y_pred = a * X**2 + b * X + c
    N = len(X)

    da = (-2/N) * np.sum((y_true - y_pred) * X**2)
    db = (-2/N) * np.sum((y_true - y_pred) * X)
    dc = (-2/N) * np.sum(y_true - y_pred)

    return da, db, dc

def gradient_descent(X, y, initial_a, initial_b, initial_c, learning_rate, epochs):
    a, b, c = initial_a, initial_b, initial_c
    losses = []

    for epoch in range(epochs):
        da, db, dc = calc_gradient_2nd_poly_for_GD(X, y, a, b, c)

        a = a - learning_rate * da
        b = b - learning_rate * db
        c = c - learning_rate * dc

        y_pred = a * X**2 + b * X + c
        loss = calculate_mse(y, y_pred)
        losses.append(loss)

    return a, b, c, losses

learning_rate = 0.0003
epochs = 30000

print("Starting Gradient Descent training...")
final_a, final_b, final_c, losses = gradient_descent(X, y, initial_a, initial_b, initial_c, learning_rate, epochs)
print("Gradient Descent training complete.")
print(f"Initial coefficients: a={initial_a:.4f}, b={initial_b:.4f}, c={initial_c:.4f}")
print(f"Final learned coefficients: a={final_a:.4f}, b={final_b:.4f}, c={final_c:.4f}")
print(f"Final MSE loss: {losses[-1]:.4f}")

fig_loss, ax_loss = plt.subplots(figsize=(10, 6))
ax_loss.plot(losses)
ax_loss.set_title('Loss over Epochs during Gradient Descent')
ax_loss.set_xlabel('Epoch')
ax_loss.set_ylabel('Mean Squared Error')
ax_loss.grid(True)
plt.show()
```

The loss curve clearly shows that the model successfully minimized the Mean Squared Error over the epochs, indicating convergence. Now, let's see how well the learned polynomial fits our noisy data.

```python
X_fit = np.linspace(X.min(), X.max(), 100)
y_fit = final_a * X_fit**2 + final_b * X_fit + final_c

fig_fit, ax_fit = plt.subplots(figsize=(10, 6))
ax_fit.scatter(X, y, label='Noisy Data', alpha=0.6)
ax_fit.plot(X_fit, y_fit, color='red', label='Fitted Polynomial (Gradient Descent)')
ax_fit.set_title('Gradient Descent: Final Fitted Polynomial')
ax_fit.set_xlabel('X')
ax_fit.set_ylabel('y')
ax_fit.legend()
ax_fit.grid(True)
plt.show()
```

The fitted curve shows a good approximation of the underlying pattern in the noisy data. Finally, let's look at the coefficients that Gradient Descent found.

```python
print(f"Final Learned Coefficients (a, b, c):\n  a = {final_a:.4f}\n  b = {final_b:.4f}\n  c = {final_c:.4f}")

print("\nGround Truth Coefficients:\n  a = 1\n  b = 0\n  c = 0")
```

As you can see, the learned coefficients are very close to the true coefficients (a=1, b=0, c=0) for the original `y = x^2` function, demonstrating the effectiveness of Gradient Descent in finding optimal model parameters even with noisy data.
